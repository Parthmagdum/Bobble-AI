```python
import pandas as pd

df = pd.read_excel("/content/drive/MyDrive/sales_data.csv.xlsx")

print(df.head())
print(df.info())
print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Sales', data=df, estimator=sum)
plt.title("Sales Distribution Across Product Categories")
plt.ylabel("Total Sales")
plt.xlabel("Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Sales', y='Profit', data=df)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Quantity'], bins=10, kde=True)
plt.title("Distribution of Quantity Sold")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Region', y='Discount', data=df)
plt.title("Discounts Given in Each Region")
plt.xlabel("Region")
plt.ylabel("Discount")
plt.show()
```

```python
plt.figure(figsize=(8,5))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
